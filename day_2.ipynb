```python
#arithmetic operators

a = 10
b = 3
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

```python
#relational operators
mark = 75
print(mark == 75)
print(mark > 50)
print(mark < 50)
```

```python
#bitwise operators

#AND
a = 10
b = 8
print(a & b)

#OR
print(5 | 3)

#XOR
print(6 ^ 7)

#<<
print(3 << 2)

#<<
print(1 >> 9)
```

```python
#logical operators
age = 24
mark = 89
print(age >= 24 and mark >= 40)
```

```python
#assignment operators
x = 90
x += 3
print(x)
```

```python
#compound operators
x = 20
x += 2
print(x)
x -= 45
print(x)
x *= 5
print(x)
x //= 9
print(x)
```

```python
#membership operators
fruits = ["apple", "banana", "orange"]
print("apple" in fruits)
```

```python
#identity operators
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)
```
