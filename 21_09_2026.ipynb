## **1. Starting with Pandas**
First, we need to load the dataset and understand the data.

```python
import pandas as pd

df = pd.read_csv("student_visualization_dataset.csv")

print(df.head(10))
```

## **3. Basic Bar Chart**
Bar chart is used for **understanding** or **comparing** categories from the dataset

**X-axis → Student Name**

**Y-axis → Marks**

```python
import matplotlib.pyplot as plt

plt.bar(df["Student_Name"], df["Marks"])

plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.xticks(rotation=90)

plt.show()
```

## **4. Line Chart**

```python
plt.plot(df["Student_ID"], df["Marks"], marker="o")

plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.title("Student Marks Trend")

plt.show()
```

# **Is a line chart the most appropriate chart for comparing individual students?**

## **5. Histogram**
## A histogram tells us how the numerical values are distributed.

```python
plt.hist(df["Marks"], bins=5)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")

plt.show()
```

How many students scored between **40–50**?

How many scored between **80–90**?

Where are most students concentrated?

## **6. Box Plot**
### A box plot helps us understand the **spread of numerical data** and **identify possible outliers**

```python
plt.boxplot(df["Marks"])

plt.ylabel("Marks")
plt.title("Distribution of Marks")

plt.show()
```

## **7. Most Important — Scatter Plot**
###Does studying more hours relate to getting higher marks?

```python
plt.scatter(df["Study_Hours"], df["Marks"])

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

**X-axis → Study Hours**

**Y-axis → Marks**

Each dot represents one student and for eg :

Student 1

Study Hours = 2

Marks = 62

## **8. Explain Positive Relationship**

## When **study hours increase**, **marks also generally increase**. This is called a *positive relationship*.

## **9. Negative Relationship**

```python
plt.scatter(df["Screen_Time_Hours"], df["Marks"])

plt.xlabel("Screen Time (Hours)")
plt.ylabel("Marks")
plt.title("Screen Time vs Marks")

plt.show()
```

### If **screen time increases** while **marks generally decrease**, we can observe a negative relationship.

## **10. Correlation**

```python
correlation = df["Study_Hours"].corr(df["Marks"])

print(correlation)
```

| Correlation | Meaning                       |
| ----------: | ----------------------------- |
|        `+1` | Perfect positive relationship |
|      `+0.7` | Strong positive relationship  |
|      `+0.3` | Weak positive relationship    |
|         `0` | No linear relationship        |
|      `-0.3` | Weak negative relationship    |
|      `-0.7` | Strong negative relationship  |
|        `-1` | Perfect negative relationship |


## **11. Correlation Matrix**
This is very useful because they can see **relationships between multiple numerical variables**.

```python
correlation_matrix = df[
    ["Study_Hours",
     "Attendance_Percent",
     "Screen_Time_Hours",
     "Sleep_Hours",
     "Marks"]
].corr()

print(correlation_matrix)
```

```python
import matplotlib.pyplot as plt

plt.imshow(correlation_matrix)

plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Matrix")

plt.show()
```
