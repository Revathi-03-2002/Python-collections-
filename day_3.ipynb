```python
#if statement
marks = 80
if marks >= 50:
    print("Pass")
marks=90
if marks>=65:
  print("pass")
```

```python
#else statement
marks = 40
if marks >= 50:
    print("Pass")
else:
    print("Fail")
```

```python
#elif statement
marks = 75
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
#if else statement
marks = 65
result = "Pass" if marks >= 50 else "Fail"
print(result)
```

```python
#list
fruits = ["apple", "banana", "orange"]
fruits[1] = "mango"
print(fruits)
```

```python
#adding two list
a=[1,2,3,4]
b=[5,6,7,8,]
c=a+b
print(c)
```

```python
#repeat
number=[1,2,3,4,5,6]
print(number*3)
```

```python
# membership in
fruits = ["apple", "banana", "orange"]
print("apple" in fruits)
print("grapes" in fruits)
```

```python
# slicing
numbers=[10,20,30,40,50]
print(numbers[::2])
```

```python
#List Comprehension
numbers = [1, 2, 3, 4, 5]
squares = [n * n for n in numbers]
print(squares)
```

```python
#list methods
#append
fruits = ["apple", "banana"]
fruits.append("orange")
print(fruits)
```

```python
#insert
fruits = ["apple", "orange"]
fruits.insert(1, "banana")
print(fruits)
```

```python
#extend
fruits = ["apple", "banana"]
fruits.extend(["orange", "mango"])
print(fruits)
```

```python
#remove
fruits = ["apple", "banana", "orange"]
fruits.remove("banana")
print(fruits)
```

```python
#clear
fruits = ["apple", "banana", "orange"]
fruits.pop(1)
print(fruits)
```

```python
#reverse
numbers = [10, 20, 30, 40]
numbers.reverse()
print(numbers)
```
